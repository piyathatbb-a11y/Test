```python
## Draw Idle Frames

idle_frames_list = []

for index, offset in enumerate([-6, 0, 6], start=1):
    frame = Image.new('RGBA', (frame_size, frame_size), (0, 0, 0, 0))
    draw = ImageDraw.Draw(frame)
    draw.ellipse([20, 20, 108, 108], fill=(10, 16, 28, 220))
    draw.ellipse([34, 30 + offset, 94, 90 + offset], fill=colors['body'])
    draw.ellipse([46, 40 + offset, 86, 80 + offset], fill=colors['core'])
    draw.polygon([(38, 52 + offset), (14, 42 + offset), (14, 70 + offset)], fill=(32, 78, 111))
    draw.polygon([(90, 52 + offset), (114, 42 + offset), (114, 70 + offset)], fill=(32, 78, 111))
    draw.line([(24, 56 + offset), (42, 56 + offset)], fill=colors['accent'], width=2)
    draw.line([(84, 56 + offset), (104, 56 + offset)], fill=colors['accent'], width=2)
    draw.ellipse([56, 46 + offset, 70, 60 + offset], fill=(236, 248, 255))
    draw.ellipse([54, 44 + offset, 72, 62 + offset], outline=(220, 240, 255), width=2)
    idle_frames_list.append(frame)
```
